```python
from ISLP import load_data

Boston = load_data("Boston")
Boston.head()
```

```python
Boston['medv']
```

(a) Estimate the population mean

```python
mu_hat = Boston['medv'].mean()

print(mu_hat)
```

(b) Estimate the standard error

```python
import numpy as np
```

```python
se_mu = Boston['medv'].std() / np.sqrt(Boston.shape[0])

print(se_mu)
```

(c) Bootstrap estimate of the standard error

```python
import numpy as np

rng = np.random.default_rng(1)

bootstrap_means = []

for i in range(10000):
    sample = rng.choice(
        Boston['medv'],
        size=len(Boston),
        replace=True
    )
    
    bootstrap_means.append(sample.mean())

bootstrap_se = np.std(bootstrap_means, ddof=1)

print(bootstrap_se)
```

(d) 95% Confidence Interval

```python
lower = mu_hat - 2 * bootstrap_se
upper = mu_hat + 2 * bootstrap_se

print(lower, upper)
```

(e) Estimate the population median

```python
mu_med = Boston['medv'].median()

print(mu_med)
```

(f) Bootstrap standard error of the median

```python
bootstrap_medians = []

for i in range(10000):
    sample = rng.choice(
        Boston['medv'],
        size=len(Boston),
        replace=True
    )
    
    bootstrap_medians.append(np.median(sample))

bootstrap_median_se = np.std(
    bootstrap_medians,
    ddof=1
)

print(bootstrap_median_se)
```

(g) Estimate the tenth percentile

```python
np.percentile(Boston['medv'], 10)
```

(h) Bootstrap standard error of the tenth percentile

```python
bootstrap_percentiles = []

for i in range(10000):
    sample = rng.choice(
        Boston['medv'],
        size=len(Boston),
        replace=True
    )
    
    bootstrap_percentiles.append(
        np.percentile(sample, 10)
    )

bootstrap_percentile_se = np.std(
    bootstrap_percentiles,
    ddof=1
)

print(bootstrap_percentile_se)
```
